```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import MinMaxScaler

class TimeSeriesDataset(Dataset):
    def __init__(self, features, time_steps, history_values, future_values):
        self.features = torch.FloatTensor(features)
        self.time_steps = torch.FloatTensor(time_steps)
        self.history_values = torch.FloatTensor(history_values)
        self.future_values = torch.FloatTensor(future_values)

    def __len__(self):
        return len(self.features)

    def __getitem__(self, idx):
        return (self.features[idx],
                self.time_steps[idx],
                self.history_values[idx],
                self.future_values[idx])

def preprocess_data(file_path, history_steps=25, future_steps=5):
    # 读取Excel文件
    df = pd.read_excel(file_path)
    df = df.dropna()

    # 提取特征和目标
    features = df.iloc[:, 1:21].values  # 2-21列作为特征
    time_steps = df.iloc[:, 21].values.reshape(-1, 1)  # 22列作为时间步
    all_values = df.iloc[:, 22:52].values  # 23-52列作为沉降值

    # 确保数据完整性
    total_steps = history_steps + future_steps

    # 对每个模型进行处理
    unique_models = []
    processed_features = []
    processed_time_steps = []
    processed_history = []
    processed_future = []

    # 获取所有唯一的模型ID
    all_models = df.iloc[:, 0].unique()

    num_points = 30  # 监测点数量

    for model_id in all_models:
        model_data = df[df.iloc[:, 0] == model_id]

        # 检查这个模型是否有足够的时间步
        if len(model_data) >= total_steps:
            unique_models.append(model_id)
            model_features = model_data.iloc[0, 1:21].values  # 取第一行特征
            model_time = model_data.iloc[0, 21]  # 取第一行时间步
            model_values = model_data.iloc[:, 22:52].values  # 所有沉降值

            processed_features.append(model_features)
            processed_time_steps.append(model_time)

            # 保持原始形状：[时间步, 监测点数]
            history_values = model_values[:history_steps, :]
            future_values = model_values[history_steps:history_steps+future_steps, :]

            processed_history.append(history_values)
            processed_future.append(future_values)

    # 转换为numpy数组
    processed_features = np.array(processed_features)
    processed_time_steps = np.array(processed_time_steps).reshape(-1, 1)
    processed_history = np.array(processed_history)  # shape: [样本数, history_steps, num_points]
    processed_future = np.array(processed_future)    # shape: [样本数, future_steps, num_points]

    # 标准化
    feature_scaler = MinMaxScaler()
    time_scaler = MinMaxScaler()
    value_scaler = MinMaxScaler()

    # 调整形状以进行标准化
    flat_history = processed_history.reshape(-1, num_points)
    flat_future = processed_future.reshape(-1, num_points)
    all_values_combined = np.vstack([flat_history, flat_future])

    processed_features = feature_scaler.fit_transform(processed_features)
    processed_time_steps = time_scaler.fit_transform(processed_time_steps)
    all_values_normalized = value_scaler.fit_transform(all_values_combined)

    # 恢复原始形状
    history_normalized = all_values_normalized[:len(flat_history)].reshape(processed_history.shape)
    future_normalized = all_values_normalized[len(flat_history):].reshape(processed_future.shape)

    return (processed_features, processed_time_steps,
            history_normalized, future_normalized,
            value_scaler)

class SimpleRNN(nn.Module):
    def __init__(self, input_size, hidden_size, num_layers, num_points, future_steps):
        super(SimpleRNN, self).__init__()
        self.hidden_size = hidden_size
        self.num_layers = num_layers
        self.num_points = num_points
        self.future_steps = future_steps

        # RNN主体结构（使用GRU）
        self.rnn = nn.GRU(
            input_size=num_points,  # 改为num_points
            hidden_size=hidden_size,
            num_layers=num_layers,
            batch_first=True,
            dropout=0.2 if num_layers > 1 else 0
        )

        # 特征和时间步处理
        self.feature_proj = nn.Linear(20, 16)  # 处理20维特征
        self.time_proj = nn.Linear(1, 8)       # 处理时间步特征

        # 输出层
        self.fc = nn.Sequential(
            nn.Linear(hidden_size + 16 + 8, 128),
            nn.ReLU(),
            nn.Linear(128, future_steps * num_points)
        )

    def forward(self, features, time_steps, history):
        # history形状: [batch_size, history_steps, num_points]
        batch_size = history.size(0)

        # 处理历史序列
        rnn_out, _ = self.rnn(history)  # 直接使用history
        last_hidden = rnn_out[:, -1, :]   # 取最后一个时间步 [batch, hidden_size]

        # 处理静态特征
        features_proj = self.feature_proj(features)  # [batch, 16]
        time_proj = self.time_proj(time_steps)       # [batch, 8]

        # 合并特征
        combined = torch.cat([last_hidden, features_proj, time_proj], dim=1)

        # 生成预测
        output = self.fc(combined)
        output = output.view(batch_size, self.future_steps, self.num_points)

        return output

class CombinedModel(nn.Module):
    def __init__(self, num_points=30, future_steps=5):
        super(CombinedModel, self).__init__()
        self.rnn_model = SimpleRNN(
            input_size=num_points,  # 改为num_points
            hidden_size=128,
            num_layers=2,
            num_points=num_points,
            future_steps=future_steps
        )

    def forward(self, features, time_steps, history):
        # 直接使用history，无需permute
        return self.rnn_model(features, time_steps, history)

def train_model(model, train_loader, val_loader, criterion, optimizer,
                num_epochs=100, device='cuda' if torch.cuda.is_available() else 'cpu'):
    model.to(device)
    best_val_loss = float('inf')

    for epoch in range(num_epochs):
        model.train()
        train_loss = 0
        for features, time_steps, history, future in train_loader:
            features = features.to(device)
            time_steps = time_steps.to(device)
            history = history.to(device)
            future = future.to(device)

            optimizer.zero_grad()
            outputs = model(features, time_steps, history)
            loss = criterion(outputs, future)
            loss.backward()
            optimizer.step()

            train_loss += loss.item()

        # 验证
        model.eval()
        val_loss = 0
        with torch.no_grad():
            for features, time_steps, history, future in val_loader:
                features = features.to(device)
                time_steps = time_steps.to(device)
                history = history.to(device)
                future = future.to(device)

                outputs = model(features, time_steps, history)
                val_loss += criterion(outputs, future).item()

        train_loss /= len(train_loader)
        val_loss /= len(val_loader)

        if val_loss < best_val_loss:
            best_val_loss = val_loss
            torch.save(model.state_dict(), 'best_model.pth')

        if (epoch + 1) % 10 == 0:
            print(f'Epoch [{epoch+1}/{num_epochs}], '
                  f'Train Loss: {train_loss:.4f}, '
                  f'Val Loss: {val_loss:.4f}')

def evaluate_model(model, test_loader, value_scaler, device):
    model.eval()
    all_predictions = []
    all_targets = []
    all_inputs = []
    all_histories = []

    with torch.no_grad():
        for features, time_steps, history, future in test_loader:
            features = features.to(device)
            time_steps = time_steps.to(device)
            history = history.to(device)

            outputs = model(features, time_steps, history)

            # 收集所有数据
            all_predictions.append(outputs.cpu().numpy())
            all_targets.append(future.numpy())
            all_inputs.append(features.cpu().numpy())
            all_histories.append(history.cpu().numpy())

    # 合并所有batch的数据
    predictions = np.vstack(all_predictions)
    targets = np.vstack(all_targets)
    inputs = np.vstack(all_inputs)
    histories = np.vstack(all_histories)

    # 反标准化
    pred_reshaped = predictions.reshape(-1, predictions.shape[-1])
    target_reshaped = targets.reshape(-1, targets.shape[-1])

    predictions_original = value_scaler.inverse_transform(pred_reshaped)
    targets_original = value_scaler.inverse_transform(target_reshaped)

    predictions_original = predictions_original.reshape(predictions.shape)
    targets_original = targets_original.reshape(targets.shape)

    return predictions_original, targets_original, inputs, histories

def save_results_to_excel(train_predictions, train_targets, test_predictions, test_targets,
                         train_inputs, test_inputs, train_histories, test_histories,
                         file_name='model_results.xlsx'):
    """保存所有结果到Excel文件"""
    print(f"\n保存结果到 {file_name}...")

    with pd.ExcelWriter(file_name, engine='openpyxl') as writer:
        # 1. 创建详细预测结果sheet
        detailed_results = []

        # 处理训练集数据
        for i in range(len(train_predictions)):
            for t in range(train_predictions.shape[1]):
                for p in range(train_predictions.shape[2]):
                    detailed_results.append({
                        'Dataset': 'Train',
                        'Sample': i + 1,
                        'Time Step': t + 1,
                        'Monitor Point': p + 1,
                        'True Value': train_targets[i, t, p],
                        'Predicted Value': train_predictions[i, t, p],
                        'Absolute Error': abs(train_predictions[i, t, p] - train_targets[i, t, p]),
                        'Relative Error (%)': abs(train_predictions[i, t, p] - train_targets[i, t, p]) /
                                            (train_targets[i, t, p] if train_targets[i, t, p] != 0 else 1) * 100
                    })

        # 处理测试集数据
        for i in range(len(test_predictions)):
            for t in range(test_predictions.shape[1]):
                for p in range(test_predictions.shape[2]):
                    detailed_results.append({
                        'Dataset': 'Test',
                        'Sample': i + 1,
                        'Time Step': t + 1,
                        'Monitor Point': p + 1,
                        'True Value': test_targets[i, t, p],
                        'Predicted Value': test_predictions[i, t, p],
                        'Absolute Error': abs(test_predictions[i, t, p] - test_targets[i, t, p]),
                        'Relative Error (%)': abs(test_predictions[i, t, p] - test_targets[i, t, p]) /
                                            (test_targets[i, t, p] if test_targets[i, t, p] != 0 else 1) * 100
                    })

        # 保存详细结果
        detailed_df = pd.DataFrame(detailed_results)
        detailed_df.to_excel(writer, sheet_name='Detailed Results', index=False)

        # 2. 创建统计指标sheet
        stats_results = []

        # 计算每个监测点的统计指标
        for dataset, preds, targs in [('Train', train_predictions, train_targets),
                                    ('Test', test_predictions, test_targets)]:
            for p in range(preds.shape[2]):
                pred_values = preds[:, :, p].flatten()
                true_values = targs[:, :, p].flatten()

                mae = np.mean(np.abs(pred_values - true_values))
                mse = np.mean((pred_values - true_values) ** 2)
                rmse = np.sqrt(mse)

                # 计算R²
                ss_tot = np.sum((true_values - np.mean(true_values)) ** 2)
                ss_res = np.sum((true_values - pred_values) ** 2)
                r2 = 1 - (ss_res / ss_tot)

                stats_results.append({
                    'Dataset': dataset,
                    'Monitor Point': p + 1,
                    'MAE': mae,
                    'MSE': mse,
                    'RMSE': rmse,
                    'R²': r2
                })

        # 保存统计指标
        stats_df = pd.DataFrame(stats_results)
        stats_df.to_excel(writer, sheet_name='Statistics', index=False)

        # 3. 创建按监测点的时间序列sheet
        for p in range(train_predictions.shape[2]):
            point_results = []

            # 训练集数据
            for i in range(len(train_predictions)):
                for t in range(train_predictions.shape[1]):
                    point_results.append({
                        'Dataset': 'Train',
                        'Sample': i + 1,
                        'Time Step': t + 1,
                        'True Value': train_targets[i, t, p],
                        'Predicted Value': train_predictions[i, t, p]
                    })

            # 测试集数据
            for i in range(len(test_predictions)):
                for t in range(test_predictions.shape[1]):
                    point_results.append({
                        'Dataset': 'Test',
                        'Sample': i + 1,
                        'Time Step': t + 1,
                        'True Value': test_targets[i, t, p],
                        'Predicted Value': test_predictions[i, t, p]
                    })

            # 保存每个监测点的结果
            point_df = pd.DataFrame(point_results)
            point_df.to_excel(writer, sheet_name=f'Point_{p+1}', index=False)

        # 4. 保存输入特征
        input_features = []

        # 训练集输入特征
        for i in range(len(train_inputs)):
            feature_dict = {f'Feature_{j+1}': train_inputs[i, j] for j in range(train_inputs.shape[1])}
            feature_dict.update({
                'Dataset': 'Train',
                'Sample': i + 1
            })
            input_features.append(feature_dict)

        # 测试集输入特征
        for i in range(len(test_inputs)):
            feature_dict = {f'Feature_{j+1}': test_inputs[i, j] for j in range(test_inputs.shape[1])}
            feature_dict.update({
                'Dataset': 'Test',
                'Sample': i + 1
            })
            input_features.append(feature_dict)

        # 保存输入特征
        input_df = pd.DataFrame(input_features)
        input_df.to_excel(writer, sheet_name='Input Features', index=False)

    print(f"结果已保存到 {file_name}")

def main():
    try:
        print("开始执行主程序...")

        # 设置随机种子
        torch.manual_seed(42)
        np.random.seed(42)

        print("正在加载和预处理数据...")
        # 数据预处理
        # 在Rui电脑 中的位置
        features, time_steps, history, future, value_scaler = preprocess_data(
            'C:/Users/Rui Zhang/Downloads/FinalTimeSeriersAnalysisData.xlsx'
        )
        # 在Colab 中的位置
        #features, time_steps, history, future, value_scaler = preprocess_data(
        #    '/content/drive/MyDrive/Tunnel Settlement Data/FinalTimeSeriersAnalysisData.xlsx'
        #)
        
        # 获取实际数据集大小和划分数据集
        n_samples = len(features)
        n_train = int(n_samples * 0.7)

        train_idx = range(n_train)
        test_idx = range(n_train, n_samples)

        print(f"总样本数: {n_samples}")
        print(f"训练集样本数: {len(train_idx)}")
        print(f"测试集样本数: {len(test_idx)}")

        # 准备数据
        train_features = features[train_idx]
        train_time = time_steps[train_idx]
        train_history = history[train_idx]
        train_future = future[train_idx]

        test_features = features[test_idx]
        test_time = time_steps[test_idx]
        test_history = history[test_idx]
        test_future = future[test_idx]

        # 创建数据加载器
        train_dataset = TimeSeriesDataset(train_features, train_time, train_history, train_future)
        test_dataset = TimeSeriesDataset(test_features, test_time, test_history, test_future)

        train_loader = DataLoader(train_dataset, batch_size=8, shuffle=True)
        test_loader = DataLoader(test_dataset, batch_size=8)
        train_loader_eval = DataLoader(train_dataset, batch_size=8, shuffle=False)

        # 初始化模型
        device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
        model = CombinedModel(num_points=30, future_steps=5).to(device)
        criterion = nn.MSELoss()
        optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

        # 训练模型
        print("\n开始训练模型...")
        train_model(model, train_loader, test_loader, criterion, optimizer)
        print("训练完成!")

        # 加载最佳模型
        model.load_state_dict(torch.load('best_model.pth', map_location=device))
        model.eval()

        # 评估训练集和测试集
        print("\n评估训练集和测试集...")
        train_predictions, train_targets, train_inputs, train_histories = evaluate_model(
            model, train_loader_eval, value_scaler, device)
        test_predictions, test_targets, test_inputs, test_histories = evaluate_model(
            model, test_loader, value_scaler, device)

        # 保存所有结果
        save_results_to_excel(
            train_predictions, train_targets,
            test_predictions, test_targets,
            train_inputs, test_inputs,
            train_histories, test_histories,
            'model_results.xlsx'
        )

        print("\n程序执行完成!")

    except Exception as e:
        print(f"程序执行出错: {str(e)}")
        import traceback
        print(traceback.format_exc())

if __name__ == "__main__":
    main()
```
